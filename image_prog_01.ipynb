```python
from PIL import Image
import requests
from io import BytesIO
from IPython.display import display  # important dans Colab

url = "https://raw.githubusercontent.com/lbreilla-hue/images_SNT/main/image.jpg"
response = requests.get(url)
img = Image.open(BytesIO(response.content))

display(img)  # fonctionne mieux que img.show() dans Colab
```
